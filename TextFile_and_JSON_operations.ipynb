```python
#Open a text file in 'Write' mode to write the given lines
f=open("textfile.txt",'w')
f.write("first line\n")
f.write("second line\n")
f.write("third line\n")
f.close()
#Open a text file in 'Read' mode to read and print the available lines-readline function
k=open("textfile.txt",'r')
print(k.readline())
print(k.readline ())
print(k.readline())
k.close()
```

```python
#Open a text file in 'Append' mide to add/append new line inputs
m=open("textfile.txt",'a')
m.write("fourt line\n")
m.write("fifth line\n")
m.close()
#Open a text file in 'Read' mode to read and print the available full text- read function
p=open("textfile.txt",'r')
print(p.read())
p.close()
```

```python
with open("textfile.txt",'r') as file:
  lines=file.readlines()
  print(lines)
```

```python
import json
#JSON string
x='{"name":"John","age":30,"city":"New York"}'
#Convert JSON string to Python dictionary
y=json.loads(x)
print(y["age"])
```

```python
#Python dictionary
data={
    "name":"John",
    "age":30,
    "city":"New York"}
#Convert Python dictionary to JSON string
json_data=json.dumps(data)
print(json_data)
```
