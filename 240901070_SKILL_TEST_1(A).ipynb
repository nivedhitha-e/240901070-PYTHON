```python
PROBLEM STATEMENT:
   You are provided with the exam scores of 20 students in a class.
Write a python program to analyze these scores.Your analysis should include:
1.OVERALL CLASS PERFOMANCE: Calculate average,highest and lowest scores.
2.PASS/FAIL CLASSIFICATION:Identify which students passed the exam
3.GRADING ON A CURVE:Assign letter grades based on curved scoring or predefined thresholds
```

```python
student_data = {
    "Alan": 85, "Ben": 42, "Cara": 78, "Daniel": 92, "Elie": 55,
    "Fran": 61, "Grace": 38, "Hannah": 89, "Ivan": 74, "Judith": 67,
    "Ken": 50, "Lily": 49, "Marshall": 95, "Nancy": 82, "Ovan": 70,
    "Penny": 63, "Quentin": 58, "Ronald": 45, "Sybil": 99, "Trent": 72
}

def analyze_class_data(data):
    scores = list(data.values())

    avg_score = sum(scores) / len(scores)
    high_score = max(scores)
    low_score = min(scores)

    print("CLASS SUMMARY")
    print(f"Average: {avg_score:.1f} | Highest: {high_score} | Lowest: {low_score}\n")

    print(f"{'Name':<10}  | {'Score':<5}  | {'Status':<6}  | {'Grade':<5}")
    print("-" * 35)

    for name, score in data.items():

        status = "PASS" if score >= 50 else "FAIL"
        if score >= 90: grade = 'A'
        elif score >= 80: grade = 'B'
        elif score >= 70: grade = 'C'
        elif score >= 60: grade = 'D'
        else: grade = 'F'

        print(f"{name:<10} | {score:<5} | {status:<6} | {grade:<5}")

analyze_class_data(student_data)
```
