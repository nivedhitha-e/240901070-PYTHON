```python
  PROBLEM STATEMENT:
   Write a python program to solve a system of linear equations represented by:
     Ax=b
  Where:
    A is a  2*2 matrix
    B is a 2*1 matrix
  The program should  compute and return the solution vector x
```

```python
import numpy as np
A = np.array([[3, 1],
              [1, 2]])
b = np.array([9, 8])
x = np.linalg.solve(A, b)
print("Matrix A:\n",A)
print("Vector b:" ,b)
print("Solution vector x:",x)
```
