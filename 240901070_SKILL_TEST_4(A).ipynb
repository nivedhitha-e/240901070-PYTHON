```python
Problem Statement:
You have collected EEG data from multiple participants while they performed a cognitive
task. Each dataset includes time-series measurements of brainwave activity across various
electrodes. Write a Python program using Pandas to:
● Load and organize the EEG data.
● Perform basic statistical analysis (mean, variance, peak values) for each participant
and electrode.
● Identify patterns or trends in brainwave activity during the task.
```

```python
import pandas as pd
import numpy as np

# Example EEG dataset
data = {
    "Participant": ["P1","P1","P1","P2","P2","P2"],
    "Electrode": ["E1","E2","E3","E1","E2","E3"],
    "Signal": [12.5, 15.2, 14.1, 13.3, 16.5, 15.0]
}

# Load data into Pandas DataFrame
df = pd.DataFrame(data)

print("EEG Data:\n", df)

# Basic statistical analysis
stats = df.groupby(["Participant","Electrode"])["Signal"].agg(
    Mean="mean",
    Variance="var",
    Peak="max"
)

print("\nStatistical Analysis:\n", stats)

# Identify trend (average brain activity per participant)
trend = df.groupby("Participant")["Signal"].mean()
print("\nAverage Brainwave Activity per Participant:\n", trend)
```
