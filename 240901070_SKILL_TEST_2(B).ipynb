```python
Problem Statement
Write a Python program using NumPy to analyze the grades of 30 students across three
subjects: Math, Science, and English. Your program should:
● Calculate overall and subject-wise statistics such as average, highest, and lowest
grades.
● Identify top-performing students based on total or average scores.
● Optionally visualize the grade distributions for each subject using histograms or bar
charts.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Grades of 30 students in Math, Science, English
grades = np.random.randint(50, 100, (30, 3))

# Subject-wise statistics
print("\nAverage marks:", np.mean(grades, axis=0))
print("Highest marks:", np.max(grades, axis=0))
print("Lowest marks:", np.min(grades, axis=0))

# Top performing student
total = np.sum(grades, axis=1)
top_student = np.argmax(total)

print("\nTop student index:", top_student)
print("Top student total marks:", total[top_student])

# Histogram for grade distribution
plt.hist(grades)
plt.title("Grade Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```
