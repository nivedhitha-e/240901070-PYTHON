```python
Problem Statement:
Write a Python program using NumPy to analyze your daily step count data collected over
the past month. The program should:
● Compute key statistics such as total steps, average steps per day, and
maximum/minimum step counts.
● Identify days with high activity (e.g., days with steps above a certain threshold).
● Analyze trends over time, such as increases or decreases in activity, and optionally
visualize them.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Step count for 30 days (example data)
steps = np.random.randint(3000, 15000, 30)

# Statistics
print("\nTotal steps:", np.sum(steps))
print("Average steps per day:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

# High activity days (above 10000 steps)
high_activity = np.where(steps > 10000)
print("\nHigh activity days:", high_activity[0] + 1)

# Analyze trend (increase or decrease)
trend = np.diff(steps)
increase_days = np.sum(trend > 0)
decrease_days = np.sum(trend < 0)

print("\nDays with increase in activity:", increase_days)
print("Days with decrease in activity:", decrease_days)

# Visualization
plt.plot(steps, marker='o')
plt.title("Daily Step Count Over a Month")
plt.xlabel("Day")
plt.ylabel("Steps")
plt.show()
```
