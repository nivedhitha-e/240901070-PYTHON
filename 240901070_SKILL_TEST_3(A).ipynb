```python
Problem Statement:
Write a Python program to calculate the total impedance of an electrical circuit using the
given impedance values of individual components—resistors, capacitors, and inductors.
Your program should be able to:
● Compute total impedance for components connected in series and parallel.
● Support impedance values in complex form (to account for frequency-dependent
elements like capacitors and inductors).
● Optionally allow the user to specify the circuit configuration and component values.
```

```python
import numpy as np
# Enter impedance values (complex numbers)
# Example components: Resistor, Inductor, Capacitor
Z1 = complex(12, 0) # Resistor (10 Ω)
Z2 = complex(0, 5) # Inductor (j5 Ω)
Z3 = complex(0, -3) # Capacitor (-j3 Ω)
# Store impedances in a list
impedances = np.array([Z1, Z2, Z3])
print("Impedances:", impedances)
# ---- SERIES IMPEDANCE ----
Z_series = np.sum(impedances)
print("\nTotal Impedance in Series:", Z_series)
# ---- PARALLEL IMPEDANCE ----
Z_parallel = 1 / np.sum(1 / impedances)
print("Total Impedance in Parallel:", Z_parallel)
# ---- OPTIONAL USER INPUT ----
choice = input("\nEnter circuit type (series/parallel): ")
if choice == "series":
    print("Total Circuit Impedance:", Z_series)
elif choice == "parallel":
    print("Total Circuit Impedance:", Z_parallel)
else:
    print("Invalid choice")
```
